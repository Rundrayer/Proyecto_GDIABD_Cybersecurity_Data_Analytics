```python
#DESCOMPRESION DE ORIGEN DE DATOS


# !!! Asegurarse de tener en la carpeta Recursos el dataset con el nombre "cybersecurity_intrusion_data" en formato .zip

#Creacion de rutas auxiliares de trabajo y descompresion de archivos

# Creacion de ruta si existe el origen de datos
if (RECURSOS / "cybersecurity_intrusion_data.zip").exists():
    Carpeta_Origen_zip = RECURSOS / "cybersecurity_intrusion_data.zip"

    #descompresion del archivo zip, del origen de datos
    if not (CARGADOS / "cybersecurity_intrusion_data.csv").exists():
        with zipfile.ZipFile(Carpeta_Origen_zip, 'r') as zip_ref:
            zip_ref.extractall(CARGADOS)
        print(f"*** Dataset principal descomprimido en: {CARGADOS} ***")
    else:
        print("El dataset ya se cargo, se omite la carga")
    
    #Creacion de rutas ya establecidas de forma interna en el origen de datos 
    Carpeta_Destino = CARGADOS / "cybersecurity_intrusion_data"
    Subcarpeta_Generate = Carpeta_Destino / "GeneratedLabelledFlows.zip"
    Subcarpeta_Mlearning = Carpeta_Destino/ "MachineLearningCSV.zip"
    
    #descompresion de subcarpetas
    if not (Carpeta_Destino / "GeneratedLabelledFlows").exists():
        with zipfile.ZipFile(Subcarpeta_Generate, 'r') as zip_ref:
            zip_ref.extractall(Carpeta_Destino)
        print(f"*** GeneratedLabelledFlows descomprimido en: {Carpeta_Destino} ***")
    else:
        print("GeneratedLabelledFlows ya estaba descomprimido, se omite la carga")
        

    if not (Carpeta_Destino / "MachineLearningCSV").exists():
        with zipfile.ZipFile(Subcarpeta_Mlearning, 'r') as zip_ref:
            zip_ref.extractall(Carpeta_Destino)
        print(f"***MachineLearningCSV  descomprimido en: {Carpeta_Destino} ***")
    else:
        print("MachineLearningCSV ya estaba descomprimido, se omite la carga")

        
else:
    print("No se ha cargado el dataset en la carpeta Recursos, o tiene problemas con la version del mismo")
```

Como siguiente paso se va a realizar una concatenacion de registros, considerando que tienen la misma estrcutura de columnas y apilarlas seria la mejor opcion para este macro dataframe de registros de ciberseguridad de intrusion 


```python
print("Concatenacion de registros para ambos conjuntos de dataframes...")

#concatenacion por medio de ciclo incluyendo todo el conjunto de dataframes dentro del mismo concat
lista_dfs_crudos = [df for nombre, df in archivos_crudos]
df_crudo_total = pd.concat(lista_dfs_crudos, ignore_index=True)

print(" Macro-DataFrame crudo de registros de intrusion creado.") 
print(f"Tamaño total: filas [{df_crudo_total.shape[0]}] y columnas [{df_crudo_total.shape[1]}]")

lista_dfs_ml = [df for nombre, df in archivos_ml]
df_ml_total = pd.concat(lista_dfs_ml, ignore_index=True)

print(" Macro-DataFrame de machine learning creado.") 
print(f"Tamaño total: filas [{df_ml_total.shape[0]}] y columnas [{df_ml_total.shape[1]}]")


```

```python
#Guardado de macro dataframes en fomato parqeu para tener mejor rendimiento

#Guardar el dataframe de Machine Learning
ruta_ml = ANALITICOS_PARQUET / "dataset_ml_unificado.parquet"
df_ml_total.to_parquet(ruta_ml, index=False)
print(f"Macro dataframe Machine learning guardado en {ruta_ml}")

#Guardar el dataframe Crudo 
ruta_crudo = PROCESADOS / "dataset_crudo_unificado.parquet"
df_crudo_total.to_parquet(ruta_crudo, index=False)
print(f"Macro dataframe crudo de insidencias guardado en {ruta_crudo}")
```
